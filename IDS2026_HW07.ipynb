The second step of CRISP-DM is data understanding. The goal is to collect, describe, explore, and verify the quality of the data to ensure it’s suitable for the analytical or modeling tasks ahead.

**a. Describe the data. Number of rows and columns, what does one row represent.**

```python
## TODO
import pandas as pd

df_rain = pd.read_csv("data_rain.csv", sep=";", decimal=",")
df_temperature = pd.read_csv("data_temperature.csv", sep=";", decimal=",")

keys = [
    'jaam_nimi - Station name, local (est)',
    'aasta - Year of measurement, UTC time',
    'kuu - Month of measurement, UTC time',
    'paev - Day of measurement, UTC time'
]

df = pd.merge(df_temperature, df_rain, on=keys, how='inner')

df = df.rename(
    columns={
        'jaam_nimi - Station name, local (est)': 'station',
        'aasta - Year of measurement, UTC time': 'year',
        'kuu - Month of measurement, UTC time': 'month',
        'paev - Day of measurement, UTC time': 'day',
        'vaartus - The measured value_x': 'temperature',  
        'vaartus - The measured value_y': 'precipitation',  
    }
)
print(df.shape)
print(df.head(5))
```

**b. Take a look at the data dictionaries or metadata to understand the meaning of each variable.**

Make sure to differenciate between technical type and meaning (integers for categorical, strings for dates, etc).

Find out the allowed values or range for each variable, check how categories are coded, and missing values represented. Think of the role of the variable in the project (target, candidate predictor, identifier, date/time, or not relevant).

If you can, try to find where the documentation and the data disagree.

You can do this as a text paragraph, a list, or as a table suggested below.


```python
import pandas as pd

table_data = []

for col in df.columns:
    tech_type = str(df[col].dtype)
    missing_count = df[col].isnull().sum()
    unique_count = df[col].nunique()
    
    # Attribution logique adaptée aux colonnes renommées et aux suffixes
    if col == "temperature":
        meaning = "Température moyenne journalière de l'air"
        unit = "°C"
        role = "Candidate Predictor"
        status = "OK"
        notes = "Température convertie proprement"
    elif col == "precipitation":
        meaning = "Cumul journalier des précipitations"
        unit = "mm"
        role = "Target"
        status = "OK"
        notes = "Précipitations converties proprement"
    elif col == "station":
        meaning = "Nom de la station météorologique en estonien"
        unit = "N/A"
        role = "Identifier / Candidate Predictor"
        status = "OK"
        notes = "Noms locaux en estonien"
    elif col in ["year", "month", "day"]:
        meaning = "Composante temporelle (UTC)"
        unit = "N/A"
        role = "Date/Time"
        status = "OK"
        notes = "Éclaté sur une colonne distincte"
    elif "element_nimi_eng" in col:
        meaning = "Nom de l'élément mesuré en anglais"
        unit = "N/A"
        role = "Not relevant"
        status = "Redundant"
        notes = "Valeur constante (duplication suite au merge)"
    elif "element_yhik_eng" in col:
        meaning = "Unité de mesure en anglais"
        unit = "N/A"
        role = "Not relevant"
        status = "Redundant"
        notes = "Redondant avec la métadonnée précédente"
    else:
        meaning = "Autre colonne"
        unit = "N/A"
        role = "Not relevant"
        status = "OK"
        notes = ""

    # Calcul sécurisé de la plage ou des valeurs uniques
    try:
        if pd.api.types.is_numeric_dtype(df[col]):
            val_range = f"Min: {df[col].min()}, Max: {df[col].max()}"
        else:
            val_range = f"{unique_count} valeurs uniques"
    except Exception:
        val_range = "N/A"

    table_data.append({
        "variable": f"`{col}`",
        "meaning": meaning,
        "unit": unit,
        "type": f"Tech: `{tech_type}`",
        "values/range": val_range,
        "missing coding": f"`NaN` ({missing_count} manquants)",
        "role": role,
        "status": status,
        "notes/concerns": notes,
    })

# 4. Création du DataFrame final
overview_df = pd.DataFrame(table_data)

print(overview_df)
```

**d. Check for data quality**

Missing values, duplicates, impossible or inconsistent values. Also watch out for outliers and anomalies that may require further investigation. Explain what you found and what you are going to do with the information.


```python
## TODO
import pandas as pd

print("Missing values")
missing_counts = df[['temperature', 'precipitation']].isnull().sum()
print(missing_counts)

print("\nValues which are in double")
duplicates = df.duplicated(subset=['station', 'year', 'month', 'day']).sum()
print(f"Lines which are in doubles : {duplicates}")

print("\nImpossible value")
neg_precip = (df['precipitation'] < 0).sum()
print(f"Negatives precipitations : {neg_precip}")

extreme_temp = df[
    (df['temperature'] < -45.0) | (df['temperature'] > 37.0)
].shape[0]
print(f"Temperatures which are outside the range (-45°C à 37°C) : {extreme_temp}")

frozen_zeros = ((df['precipitation'] == 0) & (df['temperature'] < 0)).sum()
print(
    f"Frozen zero : {frozen_zeros}"
)

print("\nOutliers detections (STATISTICS)")
high_precip_threshold = df['precipitation'].quantile(0.999)
outliers_precip = df[df['precipitation'] > high_precip_threshold]
print(
    f"Extreme threshold (99.9th percentile) : {high_precip_threshold:.2f} mm"
)
print(f"Number of days of extrems precipitations : {len(outliers_precip)}")
```
